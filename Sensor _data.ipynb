```python
df_clean = df_clean.fillna({
    "engine_temp_c":0,
    "oil_pressure_psi":0,
    "vibration_level":0
})
```

```python
# keep valid records only(Engine temp > -20)
df_clean = df_clean.filter(df_clean.engine_temp_c > -20)
```

```python
df_clean = df_clean.withColumn("timestamp", to_timestamp("timestamp"))
```

##Feature Engineering

```python
# Average wheel speed

df_feature = df_clean.withColumn("avg_wheel_speed",

expr("""
     
(
wheel_speed_fl_kph +
wheel_speed_fr_kph +
wheel_speed_rl_kph +
wheel_speed_rr_kph
)/4

""")
)
```

```python
# engine stress indicator

df_feature = df_feature.withColumn(
    "engine_stress",
    when(
        (df_feature.engine_temp_c > 105) &
        (df_feature.engine_load_percent > 80), 1).otherwise(0)
    )
```

```python
# battery risk feature

df_feature = df_feature.withColumn(
    "battery_risk",
    when(
        (df_feature.battery_voltage_v < 12) |
        (df_feature.battery_health_percent < 40), 1).otherwise(0)
)
```

```python
df_feature = df_feature.withColumn(
    "brake_severity",
    when(
        df_feature.brake_pad_wear_mm <3, "critical"
    ).otherwise("normal")
)
```

## Store as Delta Table


```python
df_feature.write.format("delta").mode("overwrite").save("/Volumes/sensor_data/data/vehicle_processed")
```

##Machine Learning Models

```python
#Select ML features

features = [

"engine_temp_c",
"engine_rpm",
"oil_pressure_psi",
"engine_load_percent",
"vibration_level",
"fuel_consumption_lph",

"battery_voltage_v",
"battery_health_percent",
"battery_temp_c",

"brake_pad_wear_mm",
"brake_temp_c",

"avg_wheel_speed",

"ambient_temp_c",
"humidity_percent"

]
```

```python
from pyspark.ml.feature import VectorAssembler
# Vectorize the features
assembler = VectorAssembler(inputCols=features, outputCol="features")

df_ml=assembler.transform(df_feature)
```

```python
train_data,test_data = df_ml.randomSplit([0.8,0.2])
```

```python
from pyspark.ml.classification import RandomForestClassifier

rf_engine = RandomForestClassifier(labelCol="engine_failure_imminent", featuresCol="features")

model_engine = rf_engine.fit(train_data)
```

```python
pred_engine = model_engine.transform(test_data)
display(pred_engine)
```

```python
from pyspark.ml.evaluation import MulticlassClassificationEvaluator

evaluator = MulticlassClassificationEvaluator(labelCol="engine_failure_imminent", metricName="accuracy")

accuracy = evaluator.evaluate(pred_engine)
print(accuracy)
```

```python
pred_engine.select(

"engine_failure_imminent",
"prediction",
"probability"

).display()
```

```python
pred_engine.groupBy(

"engine_failure_imminent",
"prediction"

).count().display()
```

```python
evaluator_precision = MulticlassClassificationEvaluator(

labelCol="engine_failure_imminent",
metricName="weightedPrecision"

)

precision = evaluator_precision.evaluate(pred_engine)

print("Precision:", precision)
```

```python
evaluator_recall = MulticlassClassificationEvaluator(

labelCol="engine_failure_imminent",
metricName="weightedRecall"

)

recall = evaluator_recall.evaluate(pred_engine)

print("Recall:", recall)
```

```python
import pandas as pd

importance = model_engine.featureImportances.toArray()

feature_importance = pd.DataFrame({

"feature": features,
"importance": importance

})

feature_importance = feature_importance.sort_values(

"importance",
ascending=False

)

display(feature_importance)
```
